# EDA — Internos (module A)
Exécuter après `python scripts/generate_synthetic_data.py`. Vérifier ici le **schéma réel**, les manquants, la distribution de la cible `requests` et le choix d'un split temporel ou groupé.

```
import sys; sys.path.insert(0, '../src')
import pandas as pd, plotly.express as px
df = pd.read_parquet('../data/processed/internos_clean.parquet')
print(df.shape); df.dtypes
```

```
df.isna().sum().loc[lambda s: s>0]
```

```
df['requests'].describe(percentiles=[.5,.9,.99])
```

```
px.histogram(df, x='requests', log_y=True, nbins=80, title='Distribution de la cible (échelle log) : asymétrique -> log1p')
```

```
df.groupby('video_type')['requests'].agg(['count','mean','median','max'])
```

```
daily = df.groupby('date')['requests'].sum().reset_index()
px.line(daily, x='date', y='requests', title='Requêtes totales par jour')
```
