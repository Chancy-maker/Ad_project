# EDA — Criteo (profils de diffusion, module B)
Nécessite `data/raw/criteo_attribution_dataset.tsv.gz`. Le `timestamp` est **relatif** : aucune date réelle.

```
import sys; sys.path.insert(0, '../src')
from ad_optimizer.under_delivery.aggregation import aggregate_campaign_daily, extract_delivery_profiles
import plotly.express as px
daily = aggregate_campaign_daily('../data/raw/criteo_attribution_dataset.tsv.gz', nrows=2_000_000)  # retirer nrows pour tout lire
prof = extract_delivery_profiles(daily)
prof.describe()
```

```
px.histogram(prof, x='cv_daily', nbins=50, title='Variabilité quotidienne (coefficient de variation) par campagne')
```

```
px.histogram(prof, x='mean_daily', log_x=True, nbins=50, title='Volume quotidien moyen par campagne')
```

Ces distributions alimentent le générateur de campagnes synthétiques. Criteo ne fournit ni objectifs contractuels ni stock : ils restent synthétiques.
